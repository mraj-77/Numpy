# Indexing using Boolean Array (Boolean Indexing)

## 1. What is Boolean Indexing?
* Array ke elements par koi condition (jaise `> 50`) lagakar `True` ya `False` values ki list generate karna aur phir unhi `True` positions ka data extract karna Boolean indexing kehlata hai.
* **Real-life Example:** Jaise ek shop par bohot saare fruits hain, aur hum sirf unhi fruits ko chunna chahte hain jinki price ₹50 se kam hai.

## 2. How it Works (Example)
* **Syntax:** `arr[arr > 5]'
* Pehle `arr > 5` check karega ki kaun-kaun sa element 5 se bada hai (wo `True` ho jayenge, baaki `False`).
* Phir main array us condition ko sirf `True` wale elements par apply karke result de dega.
* **Real-life Example:** Jaise cricket team mein se sirf un players ko select karna jinhone match mein 50 se zyada runs banaye hain.

---

## Important Points ⭐
* **Condition Based** = Yeh values ko index number se nahi, balki condition (`True`/`False`) ke base par filter karta hai.
* **Filtering** = Data me se unwanted values ko hatane (filter karne) ke liye sabse zyada use hota hai.
* **Filtered Copy** = Yeh hamesha condition ko match karne wale elements ka ek naya filtered array return karta hai.

```python
import numpy as np
```

```python
arr = np.random.randint(low = 1, high = 100, size = 20).reshape(4,5)
arr
```

```python
arr > 50
```

### indexing using boolean array

```python
arr[arr>50]
```

```python
arr[(arr > 50) & (arr%2 != 0)]
```

```python
arr[(arr > 50) & (arr%2 != 0)] = 0
arr
```
