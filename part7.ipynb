# Numpy Array Operations

## 1. Basic Operations
* **a. A - b**: Array elements me se subtraction karne ke liye[cite: 7].
* **b. b * 2**: Array ke har element ko scalar value se multiply karne ke liye[cite: 7].
* **c. a < 35**: Array par conditional check lagane ke liye (boolean array return karta hai)[cite: 7].
* **d. A * B (Element-wise mul)**: Do arrays ke corresponding elements ko aapas mein multiply karne ke liye[cite: 7].
* **e. A.dot(B)**: Matrix multiplication (dot product) ke liye[cite: 7].
* **f. A *= B**: In-place element-wise multiplication aur assignment ke liye[cite: 7].

## 2. Unary Operations
* **a. A.min() / A.max()**: Array ke minimum ya maximum element ko find karne ke liye[cite: 7].
* **b. A.sum(axis=0)**: Specific axis ke along elements ka sum nikalne ke liye[cite: 7].

## 3. Universal Functions (Ufuncs)
* **a. np.exp(B)**: Exponential values calculate karne ke liye[cite: 7].
* **b. np.sqrt()**: Square root (vargmul) nikalne ke liye[cite: 7].
* **c. np.sin(A)**: Trigonometric sine values nikalne ke liye[cite: 7].

---

## Important Points ⭐
* **Basic Operations** = Arithmetic (+, -, *, dot product) aur conditional checks[cite: 7].
* **Unary Operations** = Single array par operations jaise `min`, `max`, aur `sum`[cite: 7].
* **Universal Functions** = Mathematical functions jaise `exp`, `sqrt`, aur `sin` jo poore array par ek saath lagte hain[cite: 7].

# 4 Converting Data Type (Type Casting) in NumPy

##  Using `.astype()` Method
* Array ke elements ka data type change (convert) karne ke liye `.astype()` method ka use hota hai.
* Yeh original array ko change nahi karta, balki ek naya array return karta hai naye data type ke sath.
* **Example:** `arr_float = arr_int.astype(float)`

##  Common Data Types in NumPy
* **`int` / `int32` / `int64`**: Integer values (e.g., 10, -5) ke liye.
* **`float` / `float32` / `float64`**: Decimal values (e.g., 10.5, 3.14) ke liye.
* **`bool`**: Boolean values (`True` ya `False`) ke liye.
* **`str`**: String characters ke liye.

---

## Important Points ⭐
* **`astype()`** = Array ke data type ko badalne (convert) ke liye sabse main method hai.
* **Non-destructive** = Yeh purane array ko modify nahi karta, ek naya array deta hai.
* **Memory/Precision** = Float se int convert karne par decimal values hat jaati hain.
