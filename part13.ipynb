### generate random flotting number less than 1

```python
np.random.random()
```

### Ye random numbers ko repeatable (same) banane ke liye use hota hai.

```python
np.random.seed(1)
np.random.random()
```

```python
np.random.seed(1)
a = np.random.random()   # 0.417...

np.random.seed(1)        # seed dobara reset
b = np.random.random()   # 0.417... (a jaisa hi)

print(a)
print(b)
```

### np.random.random() vs np.random.seed(1) 

Random number deta hai (0 se 1 ke beech) --	Random generator ka starting point set karta hai
