# Broadcasting in NumPy

## 1. What is Broadcasting?
* Jab alag-alag shapes (sizes) ke do arrays ke beech mathematical operations (`+`, `-`, `*`) perform karne hote hain, toh NumPy automatically chhote array ko bade array ke size ke barabar stretch (expand) kar deta hai taaki operation ho sake. Isko broadcasting kehte hain.
* **Real-life Example:** Jaise ek shop par 5 alag-alag items ki price di gayi hai, aur aapko sabhi ki price mein ₹10 ka discount (minus) karna hai. Aapko har item ke liye alag se ₹10 nahi likhna padta, balki ek single ₹10 ki value sabhi par automatically apply ho jaati hai.

## 2. Rules of Broadcasting
* NumPy do arrays ke shapes ko right side (trailing dimensions) se compare karna shuru karta hai:
  1. Dono arrays ka dimension size **equal** hona chahiye, ya phir
  2. Unme se kisi ek array ka dimension size **1** hona chahiye.
* Agar yeh conditions match nahi hoti hain, toh NumPy error (`ValueError`) deta hai.

## 3. How it Works (Example)
* **Syntax:** `arr + 5` (Yahan array 1D hai aur `5` ek single value hai, toh broadcasting ke zariye `5` poore array ke har element mein jud jata hai).
* **2D Example:** Ek `(3, 3)` matrix mein ek `(3,)` row vector ko add karna, jisme vector automatically teeno rows par spread ho jata hai.

---

## Important Points ⭐
* **Automatic Stretching** = Chhote array ya scalar ko bade array ke shape ke anusaar expand karna.
* **Shape Matching** = Operations tabhi possible hain jab shapes compatible hon (dimensions ya toh equal hon ya 1 hon).
* **No Extra Copy** = Yeh memory mein naye arrays physically copy kiye bina efficiently calculation karta hai.

```python
import numpy as np
```

## senario 1

```python
arr1 = np.arange(8).reshape(2,4)
arr2 = np.arange(8, 16).reshape(2, 4)
print(arr1)
print(arr2)
```

```python
arr1 + arr2
```

## Senario 2

```python
arr3 = np.arange(9).reshape(3, 3)
arr4 = np.arange(3).reshape(1,3)
print(arr3, arr4)
```

```python
arr3 + arr4
```

## Rule of brodcasting

### when same shape and size then add same place element

```python
arr5 = np.arange(8).reshape(2,4)
arr6 = np.arange(8, 16).reshape(2, 4)
print(arr5)
print(arr6)
arr5+arr6
```

### If x=1 and y=n then also operation will take place (Same Dimension)

```python
arr7 = np.arange(3).reshape(1, 3)
arr8 = np.arange(12).reshape(4,3)
print(arr7)
print(arr8)
```

```python
arr7 + arr8
```

### If y=1 and x=m then also operation will take place, even if they are not of the same dimension

```python
arr9 = np.arange(4).reshape(4, 1)
arr10 = np.arange(12).reshape(4,3)
print(arr9)
print(arr10)
```

```python
arr9 + arr10
```

### If x=1 and y!=n then also operation will not take place

```python
arr11 = np.arange(3).reshape(1, 3)
arr12 = np.arange(16).reshape(4,4)
print(arr11)
print(arr12)
```

```python
arr11 + arr12
```

### If x=1 and n=1 then y==m, operation to take place

```python
arr13 = np.arange(3).reshape(1, 3)
arr14 = np.arange(3).reshape(3, 1)
print(arr13)
print(arr14)
```

```python
arr13 + arr14
```

### If x=1 and y=1, then the operation will take place no matter what

```python

arr15 = np.arange(1).reshape(1, 1)
arr16 = np.arange(20).reshape(4, 5)
print(arr15)
print(arr16)
```

```python
arr15 + arr16
```

### If they are of different dimensions

```python

arr17 = np.arange(4)
arr18 = np.arange(20).reshape(5, 4)
print(arr17)
print(arr18)
```

```python
arr17 + arr18
```
