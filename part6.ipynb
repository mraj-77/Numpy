# Indexing, Slicing and Iteration

## 1. Indexing
* Array ke kisi specific element ko uske position number (index) se access karne ke liye use hota hai (jaise `arr[0]` ya 2D array mein `arr[1, 2]`).
* Indexing hamesha `0` se start hoti hai.

## 2. Slicing
* Array ke ek bade hisse ya range ko extract (baahar nikalna) karne ke liye use hota hai.
* Iska syntax `[start:stop:step]` hota hai (jaise `arr[1:5]`).

## 3. Iteration
* Array ke saare elements par ek-ek karke loop chalane (traverse karne) ke liye use hota hai.
* Isme `for` loop ka use karke array ke har ek element ko print ya modify kiya ja sakta hai.

---

## Important Points ⭐
* **Indexing** = Single element ko find karne ke liye (`arr[index]`).
* **Slicing** = Elements ke ek tukde (range) ko nikalne ke liye (`start:stop`).
* **Iteration** = Loop ke zariye saare elements ko ek-ek karke access karna.

```python
import numpy as np
```

```python
arr1 = np.arange(24)
arr1
```

```python
arr2 = np.arange(24).reshape(6,4)
arr2
```

```python
arr = np.arange(10)
arr
```

```python
arr[-1]
```

```python
arr[2:5]
```

```python
arr2 = arr1.reshape(6,4)
arr2
```

```python
arr2[2]
```

```python
arr2[:2]
```

```python
arr2[2:5]
```

```python
arr2[:2]
```

```python
arr2[:, 2:3]
```

```python
arr2[3:5, 1:3]
```

```python
arr2[3:4, 1:3]
```

```python
for i in arr2:
    print(i)
```

```python
for i in np.nditer(arr2):
    print(i)
```
