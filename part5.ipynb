# Lists Vs Arrays

## 1. Faster
* NumPy arrays Python lists ke muqable calculation karne mein bohot fast hote hain[cite: 5].

## 2. Convenient
* Arrays ke sath data par operations perform karna aur complex math karna bohot aasaan (convenient) hota hai[cite: 5].

## 3. Less Memory
* Python lists kiCompared, NumPy arrays memory (RAM) bohot kam use karte hain[cite: 5].

---

## Important Points ⭐
* **Faster** = Lists se zyada fast execution speed hoti hai[cite: 5].
* **Convenient** = Handling aur operations karna aasan hai[cite: 5].
* **Less Memory** = Storage space kam leta hai[cite: 5].

```python
import time
```

```python
x = range(10000000)
y = range(10000000, 20000000)

start_time = time.time()

c = [(x+y) for x,y in zip(x,y)]

print(time.time() - start_time)
```

```python
a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start_time = time.time()

c = a + b

print(time.time() - start_time)
```
