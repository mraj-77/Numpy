# NumPy (Numerical Python)

## 1. What is NumPy?

**NumPy stands for Numerical Python.** It is a Python library used for numerical and scientific computing.

It provides a powerful N-dimensional array object called `ndarray`, which allows us to perform mathematical operations efficiently.

### Why NumPy?

- Fast numerical computations.
- Memory-efficient array operations.
- Supports multidimensional arrays.
- Provides mathematical and statistical functions.
- Widely used in Data Science, Machine Learning and AI.

### Import NumPy

```python
import numpy as np
```

---

## 2. Applications of NumPy

NumPy provides several powerful capabilities:

### 1. N-Dimensional Array

- Provides a powerful N-dimensional array object called `ndarray`.
- Supports 1D, 2D and higher-dimensional arrays.
- Efficiently handles large numerical datasets.

### 2. Broadcasting

- Allows mathematical operations between arrays of different shapes.
- Automatically adjusts compatible dimensions during operations.
- Avoids unnecessary manual resizing of arrays.

**Example:**

```python
a = np.array([10, 20, 30])

print(a + 5)
```

**Output:**
```text
[15 25 35]
```

### 3. C/C++ and Fortran Integration

- Provides tools to integrate Python with C, C++ and Fortran code.
- Useful for high-performance numerical computing.

### 4. Mathematical Capabilities

| Feature | Application |
|---|---|
| Linear Algebra | Matrix multiplication, inverse, determinant |
| Fourier Transform (FFT) | Signal processing and frequency analysis |
| Random Number Generation | Simulations, sampling and ML experiments |

---

## 3. N-Dimensional Array Properties

NumPy arrays have the following important properties:

### 1. Homogeneous

- NumPy arrays generally contain elements of the same data type (`dtype`).
- If different data types are provided, NumPy automatically converts them into a compatible common data type.

**Example:**

```python
arr = np.array([10, 20, 30.5])

print(arr)
print(arr.dtype)
```

**Output:**
```text
[10.  20.  30.5]
float64
```

Here, integers are automatically converted into floats.

### 2. Data Types (`dtype`)

NumPy arrays are NOT limited to numbers. They can store different types of data.

| Data Type | Description |
|---|---|
| `int` | Integer values |
| `float` | Decimal values |
| `complex` | Complex numbers |
| `bool` | True / False |
| `str` | String values |
| `object` | Python objects |

**Example:**

```python
arr = np.array([1, 2, 3])

print(arr.dtype)
```

### 3. Fixed Item Size

- NumPy arrays have a fixed item size for a given data type.
- Every element of the same array occupies the same number of bytes.
- We can check the memory occupied by one element using `itemsize`.

**Example:**

```python
arr = np.array([10, 20, 30], dtype=np.int64)

print(arr.itemsize)
```

**Output:**
```text
8
```

Each element occupies **8 bytes**.

---

## 4. Important NumPy Array Attributes

| Attribute | Description |
|---|---|
| `ndim` | Returns number of dimensions |
| `shape` | Returns array dimensions |
| `size` | Returns total number of elements |
| `dtype` | Returns data type |
| `itemsize` | Returns memory occupied by one element (bytes) |
| `nbytes` | Returns total memory occupied by array (bytes) |

**Example:**

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data Type:", arr.dtype)
print("Item Size:", arr.itemsize)
print("Total Memory:", arr.nbytes)
```
