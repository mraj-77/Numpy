# Reshaping Numpy Array

## 1. Ravel (`.ravel()`)
* Kisi multi-dimensional (2D/3D) array ko flat karke 1-D array mein convert karta hai[cite: 9].
* **Real-life Example:** Jaise ek alag-alag racks wali almari ke saare kapde nikal kar ek lambi line mein seedha rakh dena.

## 2. Reshape (`.reshape()`)
* Array ke dimensions (rows aur columns) ko bina data badle naye shape mein dhalta hai[cite: 9].
* **Real-life Example:** Jaise 12 chocolates ki ek seedhi patti ko 3 rows aur 4 columns wale box mein jama dena.

## 3. Transpose (`.transpose()` / `.T`)
* Array ki rows ko columns aur columns ko rows mein badal deta hai[cite: 9].
* **Real-life Example:** Jaise ek notebook ke page ko ghuma kar horizontal se vertical kar dena.

## 4. Stacking (`np.vstack()` / `np.hstack()`)
* Do ya do se zyada arrays ko aapas mein jod kar ek bada array banata hai[cite: 9].
* **Real-life Example:** Jaise do alag-alag eeton ki diwaro ko upar-neeche ya aaju-baaju jod kar ek badi diwar banana.

## 5. Splitting (`np.hsplit()` / `np.vsplit()`)
* Ek bade array ko tukdo mein baant kar chhote arrays mein tod deta hai[cite: 9].
* **Real-life Example:** Jaise ek poore pizza ko alag-alag slices mein kaat kar alag karna.

---

## Important Points ⭐
* **Ravel** = Multi-D array ko 1-D flat banana[cite: 9].
* **Reshape** = Rows aur columns ka structure change karna[cite: 9].
* **Transpose** = Rows aur columns ko aapas mein badalna (flip karna)[cite: 9].
* **Stacking** = Arrays ko jorna (kambine karna)[cite: 9].
* **Splitting** = Array ko todna (divide karna)[cite: 9].

```python
import numpy as np
```

```python
arr1 = np.arange(12).reshape(3,4)
arr1
```

```python
arr1.ndim
```

```python
arr1.ravel()
```

```python
arr1.transpose()
```

```python
arr2 = np.arange(10, 22).reshape(3, 4)
arr2
```

```python
np.hstack((arr1, arr2))
```

```python
np.vstack((arr1, arr2))
```
