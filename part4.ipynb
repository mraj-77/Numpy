# NumPy Array Properties & Attributes

## 1. Shape (`.shape`)
* Array ki dimensions aur rows/columns ka size batata hai (jaise `(3, 3)` ya `(5,)`).

## 2. nDim (`.ndim`)
* Array ke dimensions ki total count batata hai (jaise 1-D, 2-D, ya 3-D).

## 3. Size (`.size`)
* Array ke andar total elements kitne hain, uski sankhya (count) batata hai.

## 4. Itemsize (`.itemsize`)
* Array ka har ek element memory mein kitna space (bytes mein) le raha hai, wo batata hai.

## 5. Dtype (`.dtype`)
* Array ke andar ke elements ka data type kya hai (jaise `int64`, `float64`, etc.) wo batata hai.

## 6. astype() (`.astype()`)
* Array ke elements ka data type change (convert) karne ke liye use hota hai.

---

## Important Points ⭐
* `.shape` = Rows aur columns ka structure batata hai.
* `.ndim` = Array kitne dimension ka hai (1D/2D).
* `.size` = Total elements ki ginti.
* `.itemsize` = Ek element ka size bytes mein.
* `.dtype` = Data ka type (integer/float).
* `.astype()` = Data type ko badlne ke liye.
