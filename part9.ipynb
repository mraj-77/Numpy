# Fancy Indexing in NumPy

## 1. What is Fancy Indexing?
* NumPy array me regular slicing ki jagah **indices ki list ya array** pass karke specific elements ko ek sath select karne ka tareeqa hai.
* Isse hum kisi bhi order mein items ko access kar sakte hain.
* **Real-life Example:** Jaise ek class ke attendance register mein se roll number 5, 2, aur 9 ke students ko ek sath bula lena, chahe wo line mein kahin bhi baithe hon.

## 2. Row and Column Indexing (Columns ke liye)
* 2D arrays mein hum rows ke sath-sath **columns** ki specific positions ko bhi index list dekar access kar sakte hain.
* **Syntax:** `arr[:, [0, 2]]` (Isse saari rows ke liye 0th aur 2nd column ke elements mil jayenge).
* **Real-life Example:** Jaise ek mark sheet mein se sabhi students ke sirf 'Maths' aur 'Science' ke marks wale columns ko ek sath alag nikaal lena.

## 3. How it Works
* Di gayi indices ki list ke hisaab se elements ko extract karke naya array return karta hai.
* Jis order mein index doge, data usi sequence mein milega.

---

## Important Points ⭐
* **Index Listing** = Integers ki list/array pass karke multiple elements ek sath nikalna.
* **Column Selection** = 2D arrays mein `arr[:, [col_indices]]` likh kar specific columns select kar sakte hain.
* **New Array** = Fancy indexing hamesha data ki ek **copy** banata hai, view nahi.
