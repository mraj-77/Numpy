# Part 2 – Anaconda & First Program

## 1. Anaconda
- Python distribution for Data Science and AI.
- Includes Python, Jupyter Notebook and useful libraries.
- Conda is used for package and environment management.

## 2. Installation Steps
1. Download Anaconda for Windows (64-bit).
2. Run the installer.
3. Select Just Me.
4. Choose installation location.
5. Complete installation.
6. Open Anaconda Navigator.

## 3. Launch Jupyter Notebook
1. Open Anaconda Navigator.
2. Click Launch under Jupyter Notebook.
3. Select New → Python 3.
